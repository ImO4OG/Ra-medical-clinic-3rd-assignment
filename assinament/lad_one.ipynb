```python
student_data={
    "Name":["Imran", "Kato", "Anderw", "Wassawa", "John"],
    "Maths":[89, 98, 84, 67, 56],
    "Science":[89, 98, 84, 67, 56],
    "gender":["male", "female", "male", "male", "female",],
}
students = pd.DataFrame(student_data)
students
```
